## Part 3: Exploratory Data Analysis

EDA answers three questions before any modelling:
1. **What do I have?** — dimensions, types, missingness
2. **How is each variable distributed?** — shape, spread, outliers
3. **How are variables related?** — correlations, non-linearities

### 3.1 Overview

```python
print('Shape:', advertising.shape)
print('\nColumn types:'),
print(advertising.dtypes)
print('\nMissing values:')
print(advertising.isnull().sum())
```

- **`advertising.shape`** — returns a tuple showing the dimensions of the dataframe as `(rows, columns)`. For the Advertising dataset this will return `(200, 4)` — 200 observations and 4 variables.

- **`advertising.dtypes`** — shows the data type of each column. `float64` means decimal numbers, `int64` means whole numbers, and `object` means text. This is useful for catching columns that loaded as the wrong type, for example a numeric column that loaded as text.

- **`advertising.isnull().sum()`** — counts the number of missing values in each column. `isnull()` returns `True` for every missing value and `False` otherwise, and `sum()` adds up the `True` values. A result of 0 for all columns means the dataset is complete with no missing data.

- **`'\n'`** — a newline character that adds a blank line between the printed outputs to make them easier to read.

```python
advertising.describe().round(2)
```

- **`.describe()`** — generates a summary statistics table for every numeric column in the dataframe. For each variable it computes:
    - **`count`** — number of non-missing values
    - **`mean`** — average value
    - **`std`** — standard deviation, measuring how spread out the values are
    - **`min`** — smallest value
    - **`25%`** — first quartile, 25% of values fall below this
    - **`50%`** — median, the middle value
    - **`75%`** — third quartile, 75% of values fall below this
    - **`max`** — largest value

- **`.round(2)`** — rounds all the numbers in the table to 2 decimal places to keep the output clean and readable. Without it the numbers would show many unnecessary decimal places.

This is typically the first thing you run on a new dataset to get a quick sense of the scale, spread, and range of each variable before plotting.

### 3.2 Univariate Distributions

Before running the cell below, think about what you expect:
- What shape do you expect the TV spend distribution to take?
- What about Sales?

```python
fig, axes = plt.subplots(1, 4, figsize=(14, 3))

for ax, col in zip(axes, advertising.columns):
    ax.hist(advertising[col], bins=20, edgecolor='white')
    ax.set_title(col)
    ax.set_xlabel('Value')

axes[0].set_ylabel('Count')
plt.suptitle('Univariate Distributions — Advertising', y=1.02)
plt.tight_layout()
plt.show()
```

- **`plt.subplots(1, 4, figsize=(14, 3))`** — creates a figure with 1 row and 4 columns of subplots, giving us one panel per variable. `figsize=(14, 3)` sets the total width to 14 inches and height to 3 inches. `fig` holds the overall figure and `axes` is an array of the 4 individual plot panels.

- **`zip(axes, advertising.columns)`** — pairs each subplot panel with a column name from the dataframe. So the first panel gets `TV`, the second gets `Radio`, and so on. This avoids writing four separate plotting blocks.

- **`ax.hist(advertising[col], bins=20, edgecolor='white')`** — draws a histogram for the current column. `bins=20` splits the data into 20 equally-spaced intervals. `edgecolor='white'` adds a thin white border between bars making them easier to distinguish.

- **`ax.set_title(col)`** — adds the column name as the title of each panel.

- **`ax.set_xlabel('Value')`** — labels the x-axis of each panel as "Value".

- **`axes[0].set_ylabel('Count')`** — adds a "Count" label to the y-axis of the first panel only. Since all panels share the same y-axis meaning, labelling just the leftmost one is sufficient.

- **`plt.suptitle('Univariate Distributions — Advertising', y=1.02)`** — adds an overall title above all four panels. `y=1.02` nudges it slightly above the default position to avoid overlapping with the individual panel titles.

- **`plt.tight_layout()`** — automatically adjusts spacing between panels so nothing overlaps.

- **`plt.show()`** — renders and displays the finished figure.

Graph Interpretation

- **TV** — spending is broadly spread between 0 and 300, with a roughly uniform distribution. No strong concentration at any particular level, suggesting advertisers use a wide range of TV budgets.

- **Radio** — right-skewed, with most markets spending between 0 and 20 on radio advertising and relatively few spending above 40. Low radio spend is the norm.

- **Newspaper** — strongly right-skewed, with the majority of markets spending under 25 on newspaper advertising. A small number of markets spend well above 75, making these potential outliers worth investigating.

- **Sales** — approximately bell-shaped and centred around 10–15 units. The distribution is relatively symmetric with most markets achieving moderate sales. Very low sales (under 5) and very high sales (above 20) are rare.

**Key takeaways:**
- TV has the widest range and most uniform spread of the three predictors
- Both Radio and Newspaper are right-skewed — a few markets spend heavily while most spend little
- Sales looks the most normally distributed of the four variables, which is a useful property for regression modelling

### 3.3 Scatter Plots: Each Predictor vs Sales

```python
predictors = ['TV', 'radio', 'newspaper']
fig, axes = plt.subplots(1, 3, figsize=(13, 4))

for ax, pred in zip(axes, predictors):
    ax.scatter(advertising[pred], advertising['sales'], alpha=0.5, s=20)
    ax.set_xlabel(pred)
    ax.set_ylabel('sales')

plt.suptitle('Sales vs Each Predictor', y=1.02)
plt.tight_layout()
plt.show()

# What does the shape of each relationship suggest?
# Which predictor looks most linearly related to Sales?
```

### 3.4 Correlation Matrix

```python
corr = advertising.corr()

fig, ax = plt.subplots(figsize=(5, 4))
sns.heatmap(corr, annot=True, fmt='.2f', cmap='coolwarm',
            vmin=-1, vmax=1, linewidths=0.5, ax=ax)
ax.set_title('Correlation Matrix — Advertising')
plt.tight_layout()
plt.show()

# Which predictor correlates most strongly with Sales?
# Any predictor-predictor correlations that might be a problem later?
```

### 3.5 Pair Plot (Optional — slower to render)

```python
# Uncomment to generate — takes ~10 seconds
sns.pairplot(advertising, plot_kws={'alpha': 0.4, 's': 15})
plt.suptitle('Pair Plot — Advertising', y=1.01)
plt.show()
```
